# TIME INTELLIGENCE

## Convert timestamp to datetime

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
print(df["timestamp"].dtype)
```

## Extract date,year,month,day,hour

```python
df["date"] = df["timestamp"].dt.date
df["year"] = df["timestamp"].dt.year
df["month"] = df["timestamp"].dt.month
df["day"] = df["timestamp"].dt.day
df["hour_from_timestamp"] = df["timestamp"].dt.hour
print(
    (df["hour"] == df["hour_from_timestamp"]).all()
)
```

## Extract day of week

```python
df["day_name"] = df["timestamp"].dt.day_name()
df[["timestamp", "day_name"]].head()
```

```python
df["day_name"].unique()
```

## Create time-period category

```python
def classify_time(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df["time_period"] = df["hour"].apply(classify_time)
```

```python
df["time_period"].value_counts()
```

## traffic by time period

```python
df.groupby("time_period")["vehicle_count"].mean()
```

# KPI FOUNDATION 

## Total number of records

```python
total_records = len(df)

print("Total Records:", total_records)
```

## Total vehicles

```python
total_vehicles = df["vehicle_count"].sum()

print("Total Vehicles:", total_vehicles)
```

## Average vehicle count 

```python
average_vehicles = df["vehicle_count"].mean()

print("Average Vehicle Count:", round(average_vehicles, 2))
```

## Average speed

```python
average_speed = df["avg_speed_kmph"].mean()

print("Average Speed:", round(average_speed, 2), "km/h")
```

## Average road occupancy

```python
average_occupancy = df["occupancy_percent"].mean()

print("Average Occupancy:", round(average_occupancy, 2), "%")
```

## Average travel time index

```python
average_tti = df["travel_time_index"].mean()

print("Average Travel Time Index:", round(average_tti, 2))
```

## Total accidents

```python
total_accidents = df["accident_reported"].sum()

print("Total Reported Accidents:", total_accidents)
```

## Average AQI

```python
average_aqi = df["pollution_index_aqi"].mean()

print("Average AQI:", round(average_aqi, 2))
```

## KPI table

```python
kpi = {
    "Total Records": len(df),
    "Total Vehicles": df["vehicle_count"].sum(),
    "Average Vehicle Count": round(df["vehicle_count"].mean(), 2),
    "Average Speed (km/h)": round(df["avg_speed_kmph"].mean(), 2),
    "Average Occupancy (%)": round(df["occupancy_percent"].mean(), 2),
    "Average Travel Time Index": round(df["travel_time_index"].mean(), 2),
    "Average AQI": round(df["pollution_index_aqi"].mean(), 2),
    "Total Accidents": df["accident_reported"].sum()
}

kpi_table = pd.DataFrame(
    kpi.items(),
    columns=["KPI", "Value"]
)

kpi_table
```
